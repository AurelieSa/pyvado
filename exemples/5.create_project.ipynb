# Create Vivado project

```python
from pyvado import Pyvado
```

In this tutorial, we will create a new vivado project with Pyvado library for a Basys-3 FPGA board.

First, create a pyvado instance.

```python
pv = Pyvado()
```

Now, we can create a vivado project with the following line:

```python
pv.project.create(
  project_name="my_first_project",
  project_path=".",
  board="basys3",
  target_language="VHDL",
  simulator_language="VHDL"
)
```

You can print the installed board by using the pyvado session

```python
for board in pv.session.get_boards():
  print(board)
```
